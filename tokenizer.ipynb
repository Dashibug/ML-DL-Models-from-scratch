## Byte pair encoding для кодирования байтов Unicode UTF-8 в токены

Например сначала: *h e l l o* в байтах: *104 101 108 108 111*

BPE замечает, что последовательности вроде:

- *h + e*
- *l + l*
- *he + llo*

встречаются часто, и постепенно склеивает их:

[104, 101, 108, 108, 111] -> ["he", "llo"] -> ["hello"]

И вместо 5 позиций в Transformer идёт условно 1 токен.

```python
## пусть наш корпуск текста будет - text
text = """
low lower newest widest
low lower lower
"""

## разобьем его на символы
from collections import Counter
words = text.lower().split()
word_freqs = Counter(words)
word_freqs
```

```python
## теперь каждое слово представим как последовательность токенов
splits = {
    word: list(word) 
    for word in word_freqs
}
splits
```

```python
## и теперь самое главное: считаем частоты соседних пар
def get_pair_freqs(splits, word_freqs):
    pair_freqs = Counter()
    for word, tokens in splits.items():
        freq = word_freqs[word]
        for i in range(len(tokens) - 1):
            pair = (tokens[i], tokens[i - 1])
            pair_freqs[pair] += freq

    return pair_freqs

pair_freqs = get_pair_freqs(splits, word_freqs)
pair_freqs.most_common(10)
```

```python
def merge_pair(pair, splits):
    a, b = pair
    merged_token = a + b

    new_splits = {}

    for word, tokens in splits.items():
        new_tokens = []
        i = 0

        while i < len(tokens):
            if (
                i < len(tokens) - 1
                and tokens[i] == a
                and tokens[i + 1] == b
            ):
                new_tokens.append(merged_token)
                i += 2
            else:
                new_tokens.append(tokens[i])
                i += 1

        new_splits[word] = new_tokens

    return new_splits

pair_freqs = get_pair_freqs(
    splits,
    word_freqs
)

best_pair = pair_freqs.most_common(1)[0][0]

print("best pair:", best_pair)

splits = merge_pair(
    best_pair,
    splits
)

splits
```

## Итого

**BPE** сжимает текст в более крупные токены, чтобы последовательность была короче и модель работала эффективнее.

Теперь когда мы закодировали, надо найти самые популярные пары и смерджить их, тем самым уменьшим длину последовательности

```python
def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]): # Pythonic way to iterate consecutive elements
        counts[pair] = counts.get(pair, 0) + 1
    return counts

stats = get_stats(tokens)
print(stats)
# print(sorted(((v,k) for k,v in stats.items()), reverse=True))
```

```python
top_pair = max(stats, key=stats.get)
top_pair # самая популярная пара
```

```python
## теперь надо объединить самые частые пары в новый индекс
def merge(ids, pair, idx):
    newids = []
    i = 0
    while i < len(ids):
        if i < len(ids) - 1 and ids[i] == pair[0] and ids[i + 1] == pair[1]:
            newids.append(idx)
            i += 2
        else:
            newids.append(ids)
            i += 1
    return newids

print(merge([5, 6, 6, 7, 9, 1], (6, 7), 99))
tokens2 = merge(tokens, top_pair, 256)
print(tokens2)
print("length:", len(tokens2))
        
```

```python
def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts

def merge(ids, pair, idx):
  newids = []
  i = 0
  while i < len(ids):
    if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
      newids.append(idx)
      i += 2
    else:
      newids.append(ids[i])
      i += 1
  return newids

# ---
vocab_size = 276 # the desired final vocabulary size
num_merges = vocab_size - 256
ids = list(tokens) # copy so we don't destroy the original list

merges = {} # (int, int) -> int
for i in range(num_merges):
  stats = get_stats(ids)
  pair = max(stats, key=stats.get)
  idx = 256 + i
  print(f"merging {pair} into a new token {idx}")
  ids = merge(ids, pair, idx)
  merges[pair] = idx
```

## Итого
BPE-токенайзер обучается на корпусе текста:

Текст переводится в базовые токены — обычно UTF-8 байты.
Считаются самые частые соседние пары токенов.
Самая частая пара склеивается в новый токен.
Это повторяется много раз, пока не достигнем нужного vocab_size.

В итоге токенайзер запоминает все merge-правила, и потом применяет их к новому тексту.
vocab_size — это гиперпараметр. Чем он больше, тем больше частых последовательностей мы можем склеить, и обычно тем короче становятся sequences, но при этом растёт embedding matrix модели.

Важно отметить, что токенизатор - это полностью отдельный и независимый от LLM модуль. У него есть собственный набор текстовых данных для обучения (который может отличаться от набора данных самой LLM); на этих данных формируется словарь с использованием алгоритма Byte Pair Encoding (BPE). 

В дальнейшем токенизатор обеспечивает взаимное преобразование между исходным текстом и последовательностями токенов. Сама же LLM работает исключительно с токенами и никогда не взаимодействует с текстом напрямую.

## Теперь перейдем к encoding и decoding

После того как мы обучили BPE, у нас есть только набор merge-правил и словарь. Этого мало - токенизатор ещё должен уметь работать в обе стороны.

**Encoding** - это то что мы и сделали:

raw text
→ UTF-8 bytes
→ применяем выученные BPE merges
→ token_ids

Например:

"hello"
→ [104, 101, 108, 108, 111]
→ [256, 257]

Именно это нужно перед подачей текста в LLM.

А вот **Decoding** - обратный процесс:

token_ids
→ разворачиваем BPE-токены обратно в bytes
→ UTF-8 decode
→ raw text

Например:

[256, 257]
→ [104, 101, 108, 108, 111]
→ "hello"

## decoding

Надо понять какой будет текст, соответствующий последовательности целых чисел в диапазоне [0, vocab_size]

```python
vocab = {idx: bytes([idx]) for idx in range(256)} #создаёт базовый словарь для всех 256 байтов
for (p0, p1), idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1]

def decode(ids):
    # given ids (list of integers), return Python string
  tokens = b"".join(vocab[idx] for idx in ids)
  text = tokens.decode("utf-8", errors="replace")
  return text

print(decode([128]))
```

## Итого
При обучении BPE мы склеивали байты в более крупные token IDs. При decoding делаем обратное на уровне представления: каждому ID сопоставляем исходную последовательность байтов, соединяем их и декодируем через UTF-8.

## encoding

```python
def encode(text):
    tokens = list(text.encode("utf-8"))
    ## Среди пар, которые сейчас встречаются в тексте, 
    ## найди ту, которая была раньше всего выучена BPE.
    while len(tokens) >= 2:
        stats = get_stats(tokens)
        pair = min(stats, key=lambda p: merges.get(p, float("inf")))
        if pair not in merges:
          break # nothing else can be merged
        idx = merges[pair]
        tokens = merge(tokens, pair, idx)
    return tokens

print(encode(""))
```

«Что происходит на inference, когда приходит новая строка?» — здесь важно не сказать, что мы снова ищем самые частые пары. На inference мы уже не обучаем tokenizer, а применяем заранее выученный список merges к новому тексту.

## Итого
При обучении BPE учит как объединять на определенном корпусе текста, повторая до определенного размера слвоаря. При encoding он уже ничего не учит, а просто применяет эти правила к новому тексту и получает token IDs.

## Sentencepiece

Тоже токенизатор, точнее библиотека которая поможет токенизировать слова, можно сказать аналог tiktoken, но у него есть главное отличие - **SentencePiece** может обучаться прямо на сыром тексте, без предварительного split() по словам, поэтому хорошо подходит для разных языков, **то есть работает напрямую с Unicode кодировокй, а не байтами UTF-8 как tiktoken!**



```python
import sentencepiece as spm
# write a toy.txt file with some random text
with open("toy.txt", "w", encoding="utf-8") as f:
  f.write("SentencePiece is an unsupervised text tokenizer and detokenizer mainly for Neural Network-based text generation systems where the vocabulary size is predetermined prior to the neural model training. SentencePiece implements subword units (e.g., byte-pair-encoding (BPE) [Sennrich et al.]) and unigram language model [Kudo.]) with the extension of direct training from raw sentences. SentencePiece allows us to make a purely end-to-end system that does not depend on language-specific pre/postprocessing.")
```

```python
# train a sentencepiece model on it
# the settings here are (best effort) those used for training Llama 2
import os

options = dict(
  # input spec
  input="toy.txt",
  input_format="text",
  # output spec
  model_prefix="tok400", # output filename prefix
  # algorithm spec
  # BPE alg
  model_type="bpe",
  vocab_size=400,
  # normalization
  normalization_rule_name="identity", # ew, turn off normalization
  remove_extra_whitespaces=False,
  input_sentence_size=200000000, # max number of training sentences
  max_sentence_length=4192, # max number of bytes per sentence
  seed_sentencepiece_size=1000000,
  shuffle_input_sentence=True,
  # rare word treatment
  character_coverage=0.99995,
  byte_fallback=True,
  # merge rules
  split_digits=True,
  split_by_unicode_script=True,
  split_by_whitespace=True,
  split_by_number=True,
  max_sentencepiece_length=16,
  add_dummy_prefix=True,
  allow_whitespace_only_pieces=True,
  # special tokens
  unk_id=0, # the UNK token MUST exist
  bos_id=1, # the others are optional, set to -1 to turn off
  eos_id=2,
  pad_id=-1,
  # systems
  num_threads=os.cpu_count(), # use ~all system resources
)

spm.SentencePieceTrainer.train(**options)
```

## Vocab_size

vocab_size - важный гиперпараметр, потому что он влияет сразу и на токенизацию, **и на архитектуру самой LLM**. Его выбирают при проектировании токенизатора и модели и потом фиксируют.


Если обратиться к коду [gpt.py](https://github.com/karpathy/ng-video-lecture/blob/master/gpt.py), то сразу видно где он участвует:
То есть на входе модели:

$$ token\_id \rightarrow embedding $$

и матрица имеет размер

$$ V \times d $$

где V=vocab_size

А на выходе модель должна для следующего токена выдать логит для каждого токена vocabulary:

$$ h \rightarrow [logit_1,\ldots,logit_V] $$

поэтому lm_head тоже зависит от vocab_size.

**Но стоит запомнить!** vocab_size задаёт компромисс между длиной последовательности и размером словаря. 

Большой словарь лучше сжимает текст в токены и уменьшает длину последовательности, но увеличивает кол-во параметров и стоимость предсказания по словарю. 

Маленький словарь дешевле по параметрам, но сильнее дробит текст и делает последовательности длиннее.
